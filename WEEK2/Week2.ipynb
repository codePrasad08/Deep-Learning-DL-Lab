```python
#Q1
import torch
a=torch.tensor(2.0,requires_grad=True)
b=torch.tensor(3.0,requires_grad=True)

x=2*a+3*b
y = 5*a*a + 3*b*b*b
z = 2*x + 3*y

print("x:",x)
print("y:",y)
print("z:",z)

z.backward()

with torch.no_grad():
    dzx=2
    dzy=3
    dxa=2
    dxb=3
    dya=10*a
    dyb=9*b*b
    dza=dzx*dxa+dzy*dya
    dzb=dzx*dxb+dzy*dyb
    print("\ndz/da using AD:",a.grad) #automatic diff
    print("\ndz/da in analyticl way:",dza) #manual way
    print("The function that resulted in tensor z:",z.grad_fn) #function at z
```

```python
#Q2

import torch

w=torch.tensor(2.0,requires_grad=True)
x=torch.tensor(3.0,requires_grad=True)
b=torch.tensor(4.0,requires_grad=True)

def relu(x):
    return torch.clamp_min(x,0.0)
    #clamp_min works when the paramters w,x,b are 0
    #max(0,x) or if-else block returning tensor doesn't work

u=w*x

v=u+b
a=relu(v)

a.backward()

with torch.no_grad():
    if (a>0):
        dav=1
    else:
        dav=0
    dvu=1
    duw=x
    dux=w
    dvb=1
    daw=dav*dvu*duw

    print("\nda/dw using AD:",w.grad)
    print("\nda/dw in analytical way:",daw)
```

```python
#Q3

import torch
import numpy as np

w=torch.tensor(2.0,requires_grad=True)
x=torch.tensor(3.0,requires_grad=True)
b=torch.tensor(4.0,requires_grad=True)

def sigmoid(x):
    return 1/(1+torch.exp(-x))

def sigmoid_derivative(x):
    a=-x
    b=np.exp(a)
    c=1+b
    s=1.0/c

    dsc=(-1.0/(c**2))
    dcb=1
    dba=np.exp(a)
    dax=-1

    dsx=dsc*dcb*dba*dax
    return dsx

u=w*x
v=u+b
a=sigmoid(v)

a.backward()

with torch.no_grad():
    dav=sigmoid_derivative(v) #dav=sigmoid(v)*(1-sigmoid(v))
    dvu=1
    duw=x
    dux=w
    dvb=1
    daw=dav*dvu*duw

    print("\nda/dw using AD:",w.grad)
    print("\nda/dw in analytical way:",daw)
```

```python
#Q4

import torch

x=torch.tensor(2.0,requires_grad=True)

f=torch.exp(-(x**2)-2*x-torch.sin(x))
print("f:",f)

a=-(x**2)
b=-2*x
c=-torch.sin(x)
f=torch.exp(a+b+c)

f.backward()

with torch.no_grad():
    dfa=torch.exp(a+b+c)
    dfb=torch.exp(a+b+c)
    dfc=torch.exp(a+b+c)
    dax=-2*x
    dbx=-2
    dcx=-torch.cos(x)
    dfx=dfa*dax+dfb*dbx+dfc*dcx

    print("\ndf/dx using AD:",x.grad)
    print("\ndf/dx in analytical way:",dfx)
```

```python
#Q5

import torch

x=torch.tensor(2.0,requires_grad=True)

y=8*(x**4)+3*(x**3)+7*(x**2)+6*x+3
print("y:",y)

a=8*(x**4)
b=3*(x**3)
c=7*(x**2)
d=6*x+3
y=a+b+c+d

y.backward()

with torch.no_grad():
    dya=1
    dyb=1
    dyc=1
    dyd=1
    dax=32*(x**3)
    dbx=9*(x**2)
    dcx=14*x
    ddx=6
    dyx=dya*dax+dyb*dbx+dyc*dcx+dyd*ddx
    
    print("\ndy/dx using AD:",x.grad)
    print("\ndy/dx in analytical way:",dyx)
```
