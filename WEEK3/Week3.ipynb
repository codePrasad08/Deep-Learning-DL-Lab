```python
#Q3

import torch
import matplotlib.pyplot as plt

x = torch.tensor([5.0, 7.0, 12.0, 16.0, 20.0])
y = torch.tensor([40.0, 120.0, 180.0, 210.0, 240.0])

learning_rate = torch.tensor(0.001)

class RegressionModel:
    def __init__(self):
        self.w=torch.tensor(1.0,requires_grad=True)
        self.b=torch.tensor(1.0,requires_grad=True)
    def forward(self,x):
        return self.w*x+self.b
    def update(self):
        self.w-=learning_rate*self.w.grad
        self.b-=learning_rate*self.b.grad
    def reset_grad(self):
        self.w.grad.zero_()
        self.b.grad.zero_()
    def criterion(self,yj,y_pred):
        return (y_pred-yj)**2

loss_list=[]

model=RegressionModel()

for epoch in range(100):
    loss=0.0
    for j in range(len(x)):
        y_pred=model.forward(x[j])
        loss+=model.criterion(y[j],y_pred)
    
    loss=loss/len(x)
    loss_list.append(loss.item())

    loss.backward()

    with torch.no_grad():
        model.update()
    
    model.reset_grad()

    print("Epoch: {}, w: {}, b:{}, Loss: {}".format(epoch,model.w,model.b,loss.item()))

plt.plot(loss_list)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
#Q4

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt

class LinearData(Dataset):
    def __init__(self,x,y):
        self.x=x.view(-1,1)
        self.y=y.view(-1,1)
    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
    def __len__(self):
        return self.x.shape[0]
    
class RegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.w=nn.Parameter(torch.tensor(1.0))
        self.b=nn.Parameter(torch.tensor(1.0))
    def forward(self,x):
        return self.w*x+self.b
    
x = torch.tensor([5.0, 7.0, 12.0, 16.0, 20.0])
y = torch.tensor([40.0, 120.0, 180.0, 210.0, 240.0])

learning_rate=0.001
epochs=100

dataset=LinearData(x,y)
dataloader=DataLoader(dataset, batch_size=2,shuffle=True)

model=RegressionModel()
criterion=nn.MSELoss()
optimizer=optim.SGD(model.parameters(),lr=learning_rate)

loss_list=[]

for epoch in range(epochs):
    batch_loss=0.0
    for x_batch,y_batch in dataloader:
        y_pred=model(x_batch)

        loss=criterion(y_pred,y_batch)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()

        batch_loss+=loss.item()
    loss=batch_loss/len(dataloader)
    loss_list.append(loss)
    print("Epoch: {}, w:{},b:{}, Loss: {}".format(epoch+1,model.w.item(),model.b.item(),loss))

plt.plot(loss_list)        
plt.show()
```
